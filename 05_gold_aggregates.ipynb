```python
# Properties sold by locatinn

spark.sql("""
CREATE	OR	REPLACE	TABLE	real_estate_dev.gold.price_trends_by_location	AS
SELECT	l.city,	l.state,	p.property_type,
							date_trunc('month',	p.sold_date)	AS	sold_month,
							percentile_approx(p.sold_price,	0.5)	AS	median_sold_price,
							count(*)	AS	homes_sold
FROM	real_estate_dev.silver.properties	p
JOIN	real_estate_dev.bronze.locations	l	ON	p.location_id	=	l.location_id
WHERE	p.status	=	'Sold'
GROUP	BY	1,2,3,4
""")

# Transaction status has been completed
spark.sql("""
CREATE	OR	REPLACE	TABLE	real_estate_dev.gold.agent_performance	AS
SELECT	a.agent_id,	a.first_name,	a.last_name,	a.agency_name,
							count(t.transaction_id)	AS	deals_closed,
							sum(t.commission_amount)	AS	total_commission,
							avg(t.sale_price)	AS	avg_sale_price
FROM	real_estate_dev.silver.sales_transactions	t
JOIN	real_estate_dev.bronze.agents	a	ON	t.seller_agent_id	=	a.agent_id
WHERE	t.transaction_status	=	'Completed'
GROUP	BY	1,2,3,4
""")
```
