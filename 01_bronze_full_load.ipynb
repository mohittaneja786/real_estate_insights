```python
base	= "/Volumes/real_estate_dev/bronze/vl_real_estate"

def	load_full(path_glob,	table_name):

    df	=	(
                spark.read.option("header",True)
                .option("inferSchema",	True)
                .csv(path_glob)
                .select("*","_metadata.file_name","_metadata.file_size")
            )
    
    df.write.mode("overwrite").format("delta")\
        .option("overwriteSchema", True)\
        .saveAsTable(f"real_estate_dev.bronze.{table_name}")

    print(table_name,	df.count())

load_full(f"{base}/full_load/agents_full_load.csv",	"agents")
load_full(f"{base}/full_load/locations_full_load.csv",	"locations")
load_full(f"{base}/full_load/properties_full_load_part*.csv","properties_full")
load_full(f"{base}/full_load/sales_transactions_full_load_part*.csv","sales_transactions_full")
```
